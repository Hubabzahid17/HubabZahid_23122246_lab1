```python
print("===== INTERNATIONAL HOLIDAY PLANNER =====")

place = input("Enter the country you want to explore: ")
cash = int(input("How much money can you spend? "))

if place.lower() == "japan":
    if cash >= 200000:
        print("Japan is within your travel budget!")
        print("Top attractions: Tokyo, Mount Fuji and Kyoto.")
    else:
        print("You should increase your budget before visiting Japan.")

if place.lower() == "thailand":
    if cash >= 100000:
        print("Thailand is a good option for your budget.")
        print("Top attractions: Bangkok, Phuket and Pattaya.")
    else:
        print("Your current budget is not enough for Thailand.")

if place.lower() == "italy":
    if cash >= 175000:
        print("You can enjoy a trip to Italy!")
        print("Top attractions: Rome, Venice and Milan.")
    else:
        print("You may need more money for an Italy trip.")

else:
    print("Sorry! We don't have this country in our planner.")
    print("Try Japan, Thailand or Italy.")

print("Have a safe and enjoyable journey!")
```
