```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="")


MODEL = "models/gemini-3.5-flash"


def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Pakistan tourism.

Provide a well-structured travel plan with clear headings:

1. Destination Overview
2. Budget Travel Options (Bus, Train, Flight with approx cost in PKR)
3. Places to Visit (specific locations)
4. Food Suggestions (local dishes)
5. Money-Saving Tips

Make response detailed but easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"


def emergency_response_agent(user_input):

    prompt = f"""
You are a professional Emergency Response Agent.

Provide a well-structured emergency action plan with clear headings:

1. Assess the Threat (Identify the type and severity of the emergency)
2. Alert & Notify (Trigger alarms and contact response teams and affected people)
3. Evacuate or Shelter (Direct people to safe assembly areas or secure shelter)
4. Triage & First Aid (Provide immediate life-saving care to critical casualties)
5. Contain the Danger (Isolate hazards and secure the perimeter)

Make the response direct, clear, and actionable.

User request: {user_input}

Emergency Action Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Gemini Emergency Response Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = emergency_response_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
